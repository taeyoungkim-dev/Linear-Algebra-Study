# 선형대수 기초: 벡터 연산 (Vector Basics)

이 노트북은 벡터의 기본 연산(덧셈, 스칼라 곱, 내적, 외적, 투영 등)을 다루며, 각 연산의 수학적 수식(LaTeX)과 NumPy 구현 코드를 함께 정리한 자료입니다.

```python
import numpy as np
```

## 1. 벡터의 덧셈 (Vector Addition)
두 벡터의 덧셈은 같은 차원의 벡터 간에만 성립하며, 각 요소별(element-wise) 덧셈으로 정의됩니다.

$$ \mathbf{v} + \mathbf{w} = \begin{bmatrix} v_1 + w_1 \\ v_2 + w_2 \\ \vdots \\ v_n + w_n \end{bmatrix} $$

차원이 다를 경우 발생하는 에러도 함께 확인합니다.

```python
v = np.array([4, 5, 6])
w = np.array([10, 20, 30])
u = np.array([0, 3, 6, 9])

v_w = v + w
print('v + w =', v_w)

try:
    u_w = u + w
    print(u_w)
except ValueError as e:
    print(f"Error : {e}")
```

## 2. 브로드캐스팅 (Broadcasting)
NumPy는 차원이 다른 배열 간에도 조건을 만족하면 자동으로 형태를 맞추어 연산(브로드캐스팅)을 수행합니다.
아래는 $1 \times 3$ 크기의 배열과 그 전치 행렬($3 \times 1$) 간의 덧셈으로, 결과적으로 $3 \times 3$ 행렬이 생성됩니다.

```python
w = np.array([[10, 20, 30]])
v = np.array([v]) # 차원을 맞추기 위해 배열을 2차원으로 감쌈

# 깊은 복사(Deep copy) 관점에서 사용하기도 하는 w의 전치(Transpose)
w_t = w.T 
print('v + w_t =\n', v + w_t)
```

## 3. 스칼라-벡터 곱셈 및 덧셈 (Scalar-Vector Operations)
스칼라 $s$와 벡터 $\mathbf{a}$의 연산입니다.

$$ s\mathbf{a} = \begin{bmatrix} sa_1 \\ sa_2 \\ \vdots \\ sa_n \end{bmatrix} $$

파이썬 기본 리스트(`list`)에 곱셈을 하면 리스트 자체가 반복되지만, NumPy 배열은 스칼라가 브로드캐스팅되어 각 요소별로 연산됩니다.

```python
s = 2
a = [3, 4, 5]

# 리스트 곱셈 (반복)
print('리스트 곱셈:', s * a)

# NumPy 배열 곱셈 (스칼라 곱)
b = np.array(a)
print('NumPy 배열 곱셈:', s * b)

# NumPy 배열 스칼라 덧셈 (브로드캐스팅)
print('NumPy 배열 스칼라 덧셈:', s + b)
```

## 4. 벡터의 차원과 크기 (Vector Dimension & Magnitude/Norm)
벡터의 차원과 $L_2$ 노름(Norm)을 구합니다. 벡터의 크기(Magnitude)는 원점에서 벡터 좌표까지의 직선 거리를 의미합니다.

$$ \|\mathbf{v}\|_2 = \sqrt{\mathbf{v}^T \mathbf{v}} = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2} $$

```python
v = np.array([1, 2, 3, 7, 8, 9])

v_dim = len(v)
v_magnitude = np.linalg.norm(v)

print('Dimension:', v_dim)
print('Magnitude (Norm):', v_magnitude)
```

## 5. 벡터의 내적 (Dot Product)
내적은 두 벡터의 유사도(Similarity)를 측정하거나 공간을 매핑(Mapping)할 때 쓰이는 중요한 연산입니다. 

$$ \mathbf{v}^T \mathbf{w} = \sum_{i=1}^n v_i w_i $$

내적 연산은 다음과 같은 법칙들이 성립합니다.
1. **스칼라 결합 법칙:** $s(\mathbf{v}^T \mathbf{w}) = (s\mathbf{v})^T \mathbf{w} = \mathbf{v}^T (s\mathbf{w})$
2. **분배 법칙 (Distributive law):** $\mathbf{a}^T (\mathbf{b} + \mathbf{c}) = \mathbf{a}^T \mathbf{b} + \mathbf{a}^T \mathbf{c}$

```python
v = np.array([1, 2, 3, 4])
w = np.array([5, 6, 7, 8])

print('v dot w =', np.dot(v, w))

# 스칼라 결합 법칙 증명
s = 10
print('s * (v dot w) -> dot(s*v, w):', np.dot(s * v, w))
print('s * (v dot w) -> dot(v, w*s):', np.dot(v, w * s))

# 내적의 분배 법칙 증명
a = np.array([0, 1, 2])
b = np.array([3, 5, 8])
c = np.array([13, 21, 34])

res1 = np.dot(a, b + c)
res2 = np.dot(a, b) + np.dot(a, c)
print('\n분배 법칙 성립 (a^T(b+c) == a^Tb + a^Tc):', res1 == res2)
```

## 6. 아다마르 곱 (Hadamard Product)
행렬 곱셈이나 내적과 달리, 같은 위치의 요소끼리 일반적인 곱셈을 하는 연산(Element-wise multiplication)입니다.

$$ \mathbf{a} \odot \mathbf{b} = \begin{bmatrix} a_1 b_1 \\ a_2 b_2 \\ \vdots \\ a_n b_n \end{bmatrix} $$

```python
print('Hadamard product (a * b):', a * b)
```

## 7. 벡터의 외적 (Outer Product)
외적은 열벡터와 행벡터를 곱하여 새로운 행렬을 만듭니다. $\mathbf{a}$가 $M \times 1$, $\mathbf{b}$가 $N \times 1$ 벡터일 때, 외적 결과는 $M \times N$ 차원이 됩니다.

$$ \mathbf{a} \mathbf{b}^T = \begin{bmatrix} a_1 b_1 & a_1 b_2 & \dots & a_1 b_n \\ a_2 b_1 & a_2 b_2 & \dots & a_2 b_n \\ \vdots & \vdots & \ddots & \vdots \\ a_m b_1 & a_m b_2 & \dots & a_m b_n \end{bmatrix} $$

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

outer_prod = np.outer(a, b)
print('Outer product (a b^T):\n', outer_prod)
```

## 8. 직교 투영 (Orthogonal Projection)
어떤 벡터 $\mathbf{b}$를 다른 벡터 $\mathbf{a}$ 위로 투영(Projection)할 때, 투영된 평행 성분($\mathbf{p}$)과 $\mathbf{a}$에 직교하는 직교 성분($\mathbf{e}$)으로 분해할 수 있습니다.

1. **평행 성분 (투영된 벡터, Projection):**
$$ \mathbf{p} = \text{proj}_{\mathbf{a}} \mathbf{b} = \left( \frac{\mathbf{a}^T \mathbf{b}}{\mathbf{a}^T \mathbf{a}} \right) \mathbf{a} = \beta \mathbf{a} $$

2. **직교 성분 (Orthogonal component):**
직교 성분은 원래 투영하려던 벡터 $\mathbf{b}$에서 평행 성분을 뺀 값입니다.
$$ \mathbf{b}_{\perp} = \mathbf{b} - \mathbf{p} $$

```python
a = np.array([1, 2])
b = np.array([5, 13])  # 투영할 대상 벡터

# 1. 평행 성분 (a_parallel) = a * (a^T b / a^T a)
beta = np.dot(a, b) / np.dot(a, a)
a_parallel = a * beta

# 2. 직교 성분 (b_orthogonal)
# 수정 사항: 수학적 직교 분해 정의에 맞게 b에서 평행 성분을 뺍니다.
b_orthogonal = b - a_parallel

print('스칼라 beta:', beta)
print('평행 성분 (a_parallel):', a_parallel)
print('직교 성분 (b_orthogonal):', b_orthogonal)

# 직교 확인 (내적이 0에 수렴하는지 확인)
print('직교 확인 (내적 ≈ 0):', np.dot(a, b_orthogonal))
```
