```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1.1

v = np.array([1,2])
w = np.array([5,-6])

v_add_w = v+w
v_sub_w = v-w

plt.figure(figsize=(5,5))

plt.arrow(
    0,
    0,
    v[0],
    v[1],
    head_width =0.3,
    length_includes_head=True,
    color='black',
    label='v'
    )

plt.arrow(
    v[0],
    v[1],
    w[0],
    w[1],
    head_width = 0.3,
    length_includes_head = True,
    color = 'red',
    label = 'w'
    )

plt.arrow(
    0,
    0,
    v_add_w[0],
    v_add_w[1],
    head_width = 0.3,
    length_includes_head = True,
    color = 'yellow',
    label = 'v+w'
    )

plt.arrow(
    0,
    0,
    v_sub_w[0],
    v_sub_w[1],
    head_width = 0.3,
    length_includes_head = True,
    color = 'green',
    label = 'v-w'
    )

plt.axis([-8,8,-8,8])
plt.grid()
plt.legend()
plt.show()


```

```python
#1.2
import random
def calc_norm(v):
    return sum(v*v)**(1/2)

v1 = random.randint(-100,100)
v2 = random.randint(-100,100)
v3 = random.randint(-100,100)
v4 = random.randint(-100,100)

v = np.array([v1,v2,v3,v4])

print("calc_norm(v) = ",calc_norm(v))
print("np.linalg.norm(v) = ",np.linalg.norm(v))
```
